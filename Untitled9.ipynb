```python
import kagglehub
import os

path = kagglehub.dataset_download(
    "wasifmahmood01/custom-waste-classification-dataset"
)

print("Dataset path:", path)

print("\nFolder structure:\n")

for root, dirs, files in os.walk(path):
    level = root.replace(path, "").count(os.sep)
    indent = "    " * level

    print(indent + os.path.basename(root) + "/")

    if level >= 2:
        dirs[:] = []
```

```python
import os

train_path = os.path.join(path, "wastes", "train")
test_path = os.path.join(path, "wastes", "test")

print("TRAIN CLASSES:")
for folder in os.listdir(train_path):
    if os.path.isdir(os.path.join(train_path, folder)):
        print("  ", folder)

print("\nTEST CLASSES:")
for folder in os.listdir(test_path):
    if os.path.isdir(os.path.join(test_path, folder)):
        print("  ", folder)
```

```python
import os

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

print("========== TRAIN ==========")

for folder in sorted(os.listdir(train_path)):
    folder_path = os.path.join(train_path, folder)

    if os.path.isdir(folder_path):
        count = sum(
            1 for f in os.listdir(folder_path)
            if f.lower().endswith(image_extensions)
        )
        print(f"{folder}: {count}")

print("\n========== TEST ==========")

for folder in sorted(os.listdir(test_path)):
    folder_path = os.path.join(test_path, folder)

    if os.path.isdir(folder_path):
        count = sum(
            1 for f in os.listdir(folder_path)
            if f.lower().endswith(image_extensions)
        )
        print(f"{folder}: {count}")
```

```python
import os
import shutil

selected_classes = [
    "E-waste",
    "metal waste",
    "organic waste",
    "paper waste",
    "plastic waste"
]

clean_train = "/content/waste5/train"
clean_test = "/content/waste5/test"

os.makedirs(clean_train, exist_ok=True)
os.makedirs(clean_test, exist_ok=True)

for class_name in selected_classes:

    # TRAIN
    src_train = os.path.join(train_path, class_name)
    dst_train = os.path.join(clean_train, class_name)

    shutil.copytree(
        src_train,
        dst_train,
        dirs_exist_ok=True
    )

    # TEST
    src_test = os.path.join(test_path, class_name)
    dst_test = os.path.join(clean_test, class_name)

    shutil.copytree(
        src_test,
        dst_test,
        dirs_exist_ok=True
    )

print("✅ 5 required classes copied successfully!")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    clean_train,
    validation_split=0.20,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    clean_train,
    validation_split=0.20,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    clean_test,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

class_names = train_ds.class_names

print("Classes:", class_names)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

for images, labels in train_ds.take(1):
    for i in range(min(12, len(images))):
        plt.subplot(3, 4, i + 1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

print("✅ Dataset ready for training!")
```

```python
from tensorflow.keras.applications import MobileNetV3Small

base_model = MobileNetV3Small(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

print("✅ MobileNetV3Small loaded!")
```

```python
from tensorflow.keras import layers, models

num_classes = len(class_names)

model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),

    layers.Rescaling(1./255),

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.2),

    layers.Dense(num_classes, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("✅ Model compiled!")
```

```python
EPOCHS = 15

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)
```
